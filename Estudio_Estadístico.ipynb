# Paso 5: Limpiar los datos
Elimina espacios de mas al inicio y final  de cada texto. ejemplo " MODA  " se converte en "MODA". esto es necesario por que si tiene espacios, esta palabras no son iguales.

````python
# Seleccionar columnas de texto
columnas_texto =df.select_dtypes(include='object').columns
# limpiar cada columna
for columna in columnas_texto:
  df[columna] = df[columna].str.strip()
#mostrar reslitado de la limpieza que fue completada
print('limpieza completada')
````

# Estadistica descriptiva

La estadística descriptiva es la rama de la estadística que se encarga de recolectar, organizar, resumir, analizar y presentar un conjunto de datos numéricos o cualitativos mediante tablas, gráficos y medidas de resumen.

Su objetivo principal es describir y caracterizar el comportamiento de los datos sin realizar inferencias o generalizaciones sobre una población mayor.

# Herramientas principales

**Medidas** de tendencia central: Media, mediana y moda.

**Medidas** de dispersión: Rango, varianza, desviación estándar y coeficiente de variación.

**Medidas de posición**: Cuartiles, deciles y percentiles.

Tablas de frecuencias: Frecuencia absoluta, relativa y acumulada.

Gráficos estadísticos: Histogramas, diagramas de barras, gráficos de sectores (pastel) y diagramas de caja (boxplot).



# Paso 6: Distribución por sector
¿Qué hace? Cuenta cuantos comercios hay en cada sector y calcula el porcentaje

````python
#value_conts() cuenta las veces que aparece cada valor
# normalize = True  lo converte en porcentaje
tabla_sector =pd.DataFrame({'Cantidad' : df['Categoría / Sector'].value_counts(),'Porcentaje (%)' : (df['Categoría / Sector'].value_counts(normalize=True)*100).round(2)})

print("Distribucion por sector:")
display(tabla_sector)
````

# Paso 7: Tablas cruzadas


````python
# pd.crosstab() Crea  una tabla que cruza  dos variables
# las filas son sectores ,  las columnas son rangos de precio
crosstab =pd.crosstab(df['Categoría / Sector'],df['P15_PrecioSaaS'],margins=True)
print(" tabla cruzada ver precio por sector")
display(crosstab)
````

# Paso 8: KPIs indicarores clave



````python
# Kpi 1 : % que consedera **ALto** en impacto de pago 1-CLic
#str.contains() busca texto , .mena() calcula promedio, * 100 converte a %
kpi1 =df['P2_Impacto1Clic'].str.contains('Alto|Crítico', case=False, na=False).mean()*100
# KPI 2: % Con problemas de overselling frecuente
kpi2= df['P7_FrecOverselling'].str.contains('Frecuentemente|Diariamente', case=False, na=False).mean()*100
print(f"KPI 1: Impacto 1_clic alto|Crítico {kpi1:.2f}%")
print(f"KPI 2:Overselling Frecuente|Díario {kpi2:.2f}%")
````

# Paso 9: Mapa canales
Que hace? Convierte respuesta de texto a numero para poder hacer calculo . Ejemplo : '3 - 4 canales' se converte en 3.5

````python
# Diccionario de conversiones
mapa_canales ={'1 canal':1.0, '2 canales':2.0, '3-4 canles':3.5, '5 o mas canaes':5.0}
# aplicar las conversiones
df['NumCanales_Num'] = df['P6_NumCanales'].map(mapa_canales)
#Calcular estadisticas
print(f"Promedio de calanes: {df['NumCanales_Num'].mean():.2f}")
print(f"Mediana de calanes: {df['NumCanales_Num'].median():.2f}")
````

# Paso 10: Método de pago por sector
¿Que hace?
para cara sector,  encuentra cual es el método de pago más elegido.
1. función para obtener la moda de forma segura
2. La "moda" es el valor que más se repite

````python
print("Método de pago mas usado por sector:")
def moda_segura(x):
  m=x.mode()
  return m.iloc[0] if not m.empty else 'sin datos'

# agrupar pro sector y encontrar el método  mas usado
pago_por_sector= df.groupby('Categoría / Sector')['P3_PagoPref'].agg(moda_segura)
print("Método de pago mas usado por sector:")
print(pago_por_sector)
````

# Paso 11: Disposicion a pagar por comisión
